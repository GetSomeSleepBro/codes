```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def f(x):
    return (x+3)**2
```

```python
def grad_f(x):
    return 2*(x+3)
```

```python
def gradiant_descent(start_x = 2, learning_rate = 0.1, itr = 50, tol=1e-6):
    x = start_x
    x_history = [x]

    for i in range(itr):
        gradient = grad_f(x)
        x_new = x - learning_rate * gradient

        x_history.append(x_new)

        if abs(x_new - x) < tol:
            break

        x = x_new

    return x, f(x), x_history 
```

```python
min_x, min_y, x_steps = gradiant_descent()
print("Local minima at x =", min_x)
print("Minimum value y =", min_y)
```

```python
x_vals = np.linspace(-6, 2, 150)
y_vals = f(x_vals)
```

```python
plt.plot(x_vals, y_vals, label='y = (x+3)^2')
plt.scatter(x_steps, [f(x) for x in x_steps], color='red', label='Gradient Descent Steps')
plt.title("Gradient Descent to find Local Minima")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
