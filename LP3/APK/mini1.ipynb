```python

import threading
import time

# Normal Matrix Multiplication
def normal_multiply(A, B):
    n = len(A)
    C = [[0]*n for i in range(n)]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    return C


# Function to multiply one row
def multiply_row(A, B, C, row):
    n = len(A)

    for j in range(n):
        for k in range(n):
            C[row][j] += A[row][k] * B[k][j]


# Multithreaded Matrix Multiplication
def threaded_multiply(A, B):
    n = len(A)
    C = [[0]*n for i in range(n)]
    threads = []

    for i in range(n):
        t = threading.Thread(target=multiply_row,
                             args=(A, B, C, i))
        threads.append(t)
        t.start()

    for t in threads:
        t.join()

    return C


# Main program
n = int(input("Enter matrix size: "))

print("Enter first matrix:")
A = [list(map(int, input().split())) for i in range(n)]

print("Enter second matrix:")
B = [list(map(int, input().split())) for i in range(n)]

start = time.time()
C1 = normal_multiply(A, B)
normal_time = time.time() - start

start = time.time()
C2 = threaded_multiply(A, B)
thread_time = time.time() - start

print("\nNormal Matrix Multiplication:")
for row in C1:
    print(row)

print("\nMultithreaded Matrix Multiplication:")
for row in C2:
    print(row)

print("\nNormal Time:", normal_time)
print("Threaded Time:", thread_time)
```
