```python

import heapq

def huffman(data):
    freq = {}

    # Count frequency
    for ch in data:
        freq[ch] = freq.get(ch, 0) + 1

    # Create min heap
    heap = [[f, ch] for ch, f in freq.items()]
    heapq.heapify(heap)

    # Build Huffman Tree
    while len(heap) > 1:
        a = heapq.heappop(heap)
        b = heapq.heappop(heap)

        heapq.heappush(heap, [a[0] + b[0], [a, b]])

    # Generate codes
    def generate(node, code=""):
        if isinstance(node[1], str):
            print(node[1], ":", code)
        else:
            generate(node[1][0], code + "0")
            generate(node[1][1], code + "1")

    generate(heap[0])


# Main
data = input("Enter string: ")

print("Huffman Codes:")
huffman(data)
```
