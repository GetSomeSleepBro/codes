```python
# Pass2 Assembler in Python

class ST:
    def __init__(self, index, symbol, address):
        self.index = index
        self.symbol = symbol
        self.address = address

class LT:
    def __init__(self, index, literal, address):
        self.index = index
        self.literal = literal
        self.address = address


def main():
    # ----------------- SYMBOL TABLE -----------------
    symbol_table = {
        0: ST(0, "ONE", 205),
        1: ST(1, "TWO", 209),
        2: ST(2, "RESULT", 210)
    }

    # ----------------- LITERAL TABLE ----------------
    literal_table = {
        0: LT(0, "='5'", 211),
        1: LT(1, "='3'", 212)
    }

    # ----------------- INTERMEDIATE CODE -------------
    intermediate_code = [
        "AD 01 C 200",
        "IS 04 RG 1 S 0",
        "IS 01 RG 2 S 1",
        "IS 01 RG 2 L 0",
        "IS 05 RG 3 S 2",
        "IS 01 RG 2 L 1",
        "DL 01 C 4",
        "DL 02 C 2",
        "DL 01 C 1",
        "AD 02",
        "DL 02 C 5",
        "DL 02 C 3"
    ]

    print("Final Machine Code:")
    print("-------------------")

    # ---- Generate Machine Code ----
    for line in intermediate_code:
        tokens = line.replace(",", " ").split()

        token_opcodeClass = tokens[0] if len(tokens) > 0 else None
        token_opcode = tokens[1] if len(tokens) > 1 else None
        token_operand1Type = tokens[2] if len(tokens) > 2 else None
        token_operand1 = tokens[3] if len(tokens) > 3 else None
        token_operand2Type = tokens[4] if len(tokens) > 4 else None
        token_operand2 = tokens[5] if len(tokens) > 5 else None

        # ---- Imperative Statements ----
        if token_opcodeClass == "IS":
            print(token_opcode, end=" ")

            # Operand 1: Register
            if token_operand1Type == "RG":
                print(token_operand1, end=" ")

            # Operand 2: Symbol
            if token_operand2Type == "S":
                sym = symbol_table[int(token_operand2)]
                print(sym.address, end="")

            # Operand 2: Literal
            if token_operand2Type == "L":
                lit = literal_table[int(token_operand2)]
                print(lit.address, end="")

            print()

        # ---- Declarative Statements ----
        elif token_opcodeClass == "DL":
            if token_opcode == "01":  # DC
                print("00 0", token_operand1)
            elif token_opcode == "02":  # DS
                print("00 0", token_operand1)

        # ---- Assembler Directives (AD) ----
        elif token_opcodeClass == "AD":
            continue

    # ----------------- SYMBOL TABLE PRINT -----------------
    print("\nSYMBOL TABLE:")
    for i, sym in symbol_table.items():
        print(f"{sym.index}\t{sym.symbol}\t{sym.address}")

    # ----------------- LITERAL TABLE PRINT ----------------
    print("\nLITERAL TABLE:")
    for i, lit in literal_table.items():
        print(f"{lit.index}\t{lit.literal}\t{lit.address}")


if __name__ == "__main__":
    main()
```
